```python
import os
import zipfile
import shutil
import base64
from IPython.display import HTML

exp_path = '/kaggle/working/train_'  # Adjust this path if needed

if os.path.exists(exp_path):
    zip_filename = 'trained_yolov5_model.zip'
    with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(exp_path):
            for file in files:
                file_path = os.path.join(root, file)
                arcname = os.path.relpath(file_path, exp_path)
                zipf.write(file_path, arcname)
    print("You can now download this ZIP file from Kaggle!")
    
    # List contents of the ZIP
    with zipfile.ZipFile(zip_filename, 'r') as zipf:
        print(f"\nZIP contents:")
        for file in zipf.namelist():
            print(f"  - {file}")
    # Read the ZIP file and encode it
    with open(zip_filename, 'rb') as f:
        zip_data = f.read()
    
    # Create a download link
    b64 = base64.b64encode(zip_data).decode()
    download_link = f'''
    <div style="text-align: center; padding: 20px; border: 3px solid #4CAF50; border-radius: 15px; margin: 20px; background-color: #f0f8f0;">
        <h2 style="color: #4CAF50; margin-bottom: 20px;">�� YOUR TRAINED MODEL IS READY!</h2>
        <p style="font-size: 18px; margin-bottom: 20px;">Click the button below to download your trained YOLOv5 model:</p>
        <a href="data:application/zip;base64,{b64}" download="{zip_filename}" 
           style="background-color: #4CAF50; color: white; padding: 20px 40px; text-decoration: none; 
                  font-size: 20px; border-radius: 10px; display: inline-block; margin: 10px; font-weight: bold;">
           📥 DOWNLOAD {zip_filename}
        </a>
        <p style="color: #666; margin-top: 15px;">File size: {os.path.getsize(zip_filename) / (1024*1024):.2f} MB</p>
        <p style="color: #888; font-size: 14px;">This will download your complete trained model with weights, results, and visualizations</p>
    </div>
    '''
    
    # Display the download link
    display(HTML(download_link))
  
    js_code = f'''
    <script>
    console.log("🚀 Auto-download script started");
    setTimeout(function() {{
        var link = document.createElement('a');
        link.href = '{zip_filename}';
        link.download = '{zip_filename}';
        link.style.display = 'none';
        document.body.appendChild(link);
        link.click();
        document.body.removeChild(link);
        console.log("✅ Download link clicked automatically");
    }}, 2000);
    </script>
    '''
    
    display(HTML(js_code))
```
